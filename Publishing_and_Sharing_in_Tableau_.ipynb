Q1. Tableau Public vs Tableau Server/Cloud

-   The key difference is accessibility and security:

Tableau Public: Dashboards are generally publicly accessible and published data/workbooks can be downloaded by users depending on settings. It should not be used for confidential or sensitive data.
Tableau Server/Cloud: Provides controlled, authenticated access. Administrators can manage users, groups, permissions, and data access, making it suitable for private business data.


2. Live Connection vs Extract
-  Live Connection: Tableau connects directly to the original database. Data is queried from the source whenever the dashboard needs data.
Extract: Tableau creates a stored snapshot of the data in an extract file, which can be refreshed periodically.



3. Effect on Data Refresh and Performance

-  Live connection:

Data can be more up-to-date because it comes directly from the source.
Performance depends on the database and network.
Complex queries can be slower.

Extract:

Usually provides faster dashboard performance because Tableau queries the optimized extract.
Data is only as current as the last extract refresh.
Scheduled refreshes can be configured in Tableau Server/Cloud.


4. Controlling Who Can View, Interact With, or Edit

-   The main feature is Permissions.

Tableau Server/Cloud administrators can use user/group permissions to control capabilities such as:

View
Interact
Download
Edit
Share

These permissions can be applied to workbooks, views, projects, and related content.

5. Prevent Users From Downloading Underlying Data

-   In Tableau Server, adjust the Download Data permission/capability.

Depending on the required restriction, administrators can disable capabilities such as:

Download Full Data
Download Summary Data
Download Workbook

This allows users to view the dashboard while restricting access to the underlying data.

6. Two Methods to Embed Tableau Dashboards

-  Two common methods are:

Embed code using an <iframe> in a web page.
Tableau JavaScript API / Embedding API to embed and interact with a Tableau visualization.


7. Two Best Practices for Public Sharing
-  Remove or anonymize sensitive information such as personally identifiable information (PII), patient names, phone numbers, addresses, IDs, etc.
Review permissions and download settings and ensure confidential data is not exposed through downloadable workbooks or underlying data.

For public dashboards, use only data that is appropriate for public disclosure.

8. “Data Not Found” Error After Publishing

-  A common cause is that the published workbook cannot access the original data source.

For example:

The workbook uses a local Excel/CSV file.
That file is not available to Tableau Server/Cloud.
The data source path has changed.
Required credentials have not been configured.

Solution: Ensure the data source is properly published/connected and that Tableau Server/Cloud has the required access and credentials.

9. Security Measures for an Embedded Internal Dashboard

-  Two important measures are:

Authentication: Require users to authenticate through the organization's identity system, such as SSO, before accessing the Tableau content.
Authorization/Permissions: Configure Tableau permissions and user/group access so that only authorized employees can view or interact with the embedded dashboard.

For sensitive internal dashboards, also use HTTPS for secure communication.
